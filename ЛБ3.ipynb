Напишите функцию для вычисления суммы квадратов натуральных чисел от a до b, где a и b – натуральные числа. Продемоснстрируйте использование объявленной функции.

Решение
Объявление функции
Напишите функцию, которая печатает таблицу значений функции
𝑦 = 𝑥2 + 3 ∙ 𝑥3 для 𝑥 ∈ [0, 10]. Таблица печатает значения только
для целых значений 𝑥 с шагом 1

```python
def print_values_table():

    print(f"{'x':^5} | {'y':^10}")
    print("-" * 18)

    for x in range(0, 11):

        y = x**2 + 3 * (x**3)


        print(f"{x:^5} | {y:^10}")

print_values_table()

```

Функция с возвратом значения передает результат своей работы обратно в вызвавшую её программу (с помощью ключевого слова return), а функция без возврата значения (void-функция) просто выполняет блок кода и возвращает None (в Python). Функция с возвратом может использоваться внутри выражений, например: y = math.sqrt(16).
2. Библиотечная функция — это встроенная или готовая функция, которая поставляется в составе стандартной библиотеки языка или сторонних модулей. Их не нужно писать с нуля, достаточно импортировать нужный модуль (например, import math).
3. Они похожи на «черные ящики», потому что программисту важен только вход (аргументы) и выход (результат), а внутренняя логика и устройство кода функции скрыты. Вам не нужно знать, по какому именно алгоритму вычисляется синус, чтобы использовать math.sin().
4. Инструкция генерирует случайное целое число в диапазоне от 1 до 100 включительно и сохраняет его в переменную x.

6. Она инициализирует генератор случайных чисел фиксированным числом (зерном), чтобы сделать последовательность «случайных» чисел предсказуемой и повторяемой при каждом запуске программы. Это необходимо для тестирования и отладки кода.
7. Они позволяют написать фрагмент кода один раз, а затем вызывать его в любых местах программы по имени. Если логику нужно изменить, это делается в одном месте — внутри тела функции.
8. Определение функции состоит из заголовка и тела:

	• Тело (Body): Блок кода с отступом, который выполняет основную задачу функции.

10. Рекурсия — это процесс, при котором функция вызывает саму себя внутри своего же тела. Чтобы рекурсия не стала бесконечной, в ней обязательно должно быть заложено базовое условие для выхода.
